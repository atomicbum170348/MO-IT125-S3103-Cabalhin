```python
# Load datasets
final_cleaned_data1 = pd.read_csv('/Users/fritzcabalhin/Desktop/2. Machine Learning/Codes/eda/customer_demographics_final_cleaned.csv')
final_cleaned_data2 = pd.read_csv('/Users/fritzcabalhin/Desktop/2. Machine Learning/Codes/eda/customer_transactions_final_cleaned.csv')
final_cleaned_data3 = pd.read_csv('/Users/fritzcabalhin/Desktop/2. Machine Learning/Codes/eda/social_media_interactions_final_cleaned.csv')

# Perform an outer join on the datasets using 'CustomerID'
merged1 = pd.merge(final_cleaned_data1, final_cleaned_data2, on='CustomerID', how='outer', indicator=True)
merged2 = pd.merge(merged1, final_cleaned_data3, on='CustomerID', how='outer', indicator='Merge_Status')

# Records with complete matches
final_combined_dataset = merged2.dropna()

# Records with missing values (indicating no complete match)
final_incomplete_record = merged2[merged2.isnull().any(axis=1)]

# Save the combined dataset to a CSV file
final_combined_dataset.to_csv('/Users/fritzcabalhin/Desktop/2. Machine Learning/Codes/final cleaned/final_combined_dataset.csv', index=False)

# Save the missing dataset to a CSV file
final_incomplete_record.to_csv('/Users/fritzcabalhin/Desktop/2. Machine Learning/Codes/final cleaned/final_incomplete_records.csv', index=False)

print("Data merging completed.")
print(f"Combined dataset saved with {final_combined_dataset.shape[0]} records.")
print(f"Incomplete records saved with {final_incomplete_record.shape[0]} records.")
```

```python
import pandas as pd
import plotly.express as px

# Define a custom color palette with darker shades
custom_colors = ["#006400", "#228B22", "#32CD32"]  # Darker greens for Low, Medium, High

# Ensure IncomeLevel is ordered correctly
final_combined_dataset["IncomeLevel"] = pd.Categorical(
    final_combined_dataset["IncomeLevel"], categories=["Low", "Medium", "High"], ordered=True
)

# Calculate the total count of customers per IncomeLevel
income_counts = final_combined_dataset["IncomeLevel"].value_counts().reset_index()
income_counts.columns = ["IncomeLevel", "Count"]

# Merge the total counts with the original dataset
final_combined_dataset = final_combined_dataset.merge(income_counts, on="IncomeLevel", how="left")

# Income Level Distribution
fig_income = px.bar(
    final_combined_dataset,
    x="IncomeLevel",
    color="IncomeLevel",
    title="Income Level of Customers",
    color_discrete_sequence=custom_colors,
    text="Count",  # Use the merged count column for text labels
)

# Add total count label to each bar in the graph (above the bar)
fig_income.update_traces(textposition="outside", texttemplate="%{text}")

# Show the count in the tooltip
fig_income.update_traces(
    hovertemplate="Income Level: %{x}<br>Count: %{y}<extra></extra>"  # Custom tooltip to show count
)

# Layout customization
fig_income.update_layout(
    xaxis_title="Income Level",
    yaxis_title="Count",
    plot_bgcolor="white",
    title_font=dict(size=18, family="Arial", color="darkgreen"),
)

fig_income.show()
```

```python
# Top 10 Locations by Number of Customers
location_counts = (
    final_combined_dataset.groupby("Location")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="Customer Count")
)

# Create the bar plot
fig_top_locations_customers = px.bar(
    location_counts,
    x="Customer Count",
    y="Location",
    orientation="h",
    title="Top 10 Locations by Number of Customers",
    color="Customer Count",
    color_continuous_scale=px.colors.sequential.Greens,
    text="Customer Count",  # Add the customer count as text labels on the bars
)

# Update layout to display the count labels outside the bars
fig_top_locations_customers.update_traces(
    textposition="outside",  # Position the labels outside the bars
    texttemplate="%{text}"    # Display the exact count value
)

# Layout customization
fig_top_locations_customers.update_layout(
    xaxis_title="Number of Customers",
    yaxis_title="Location",
    plot_bgcolor="white",
    title_font=dict(size=18, family="Arial", color="green"),
)

fig_top_locations_customers.show()
```

```python
import altair as alt
import pandas as pd

# Ensure the data is in the correct format to avoid warnings
final_combined_dataset['Platform'] = final_combined_dataset['Platform'].astype('category')
final_combined_dataset['InteractionType'] = final_combined_dataset['InteractionType'].astype('category')

# Calculate the total count of InteractionType per Platform
heatmap = alt.Chart(final_combined_dataset).mark_rect().encode(
    x=alt.X("Platform:N", title="Social Media Platform"),
    y=alt.Y("InteractionType:N", title="Type of Interaction"),
    color=alt.Color("count():Q", scale=alt.Scale(scheme="greens"), title="Count"),
    tooltip=["Platform", "InteractionType", "count()"],
).properties(
    title="Customer Engagement by Platform and Interaction Type",
    width=600,
    height=400,
)

# Add the total count labels on the heatmap cells
text_labels = alt.Chart(final_combined_dataset).mark_text(
    align='center',
    baseline='middle',
    fontSize=12,
    color='black'
).encode(
    x=alt.X("Platform:N"),
    y=alt.Y("InteractionType:N"),
    text=alt.Text("count():Q")
)

# Combine heatmap and text labels
final_heatmap = heatmap + text_labels

final_heatmap.interactive()
```

# Outliers to Monitor

```python
import plotly.express as px
import pandas as pd

# Create a custom color palette for Male and Female (darker shades)
color_palette = ['#006400', '#228B22']  # Darker green shades for male and female

# Create the box plot with the updated color and labels
fig_box = px.box(
    final_combined_dataset,
    x="IncomeLevel",
    y="Amount",
    color="Gender",
    title="Spending Outliers by Income Level",
    color_discrete_sequence=color_palette,
    category_orders={"IncomeLevel": ["Low", "Medium", "High"]}  # Ordering Income Levels
)

# Add count labels for Income Level per Gender
# Group by IncomeLevel and Gender to get the counts
count_data = final_combined_dataset.groupby(['IncomeLevel', 'Gender']).size().reset_index(name='Count')

# Add the count labels to the box plot
fig_box.update_traces(
    boxmean="sd",  # Show the mean and standard deviation on the box plot
    jitter=0.5,  # Adjust the spread of the points
)

# Add text labels for counts, positioned appropriately inside the box plots
for i in range(len(count_data)):
    income_level = count_data['IncomeLevel'][i]
    gender = count_data['Gender'][i]
    count = count_data['Count'][i]
    
    # Add the text label at the center of the box plot for each (IncomeLevel, Gender)
    fig_box.add_annotation(
        x=income_level,
        y=0,  # Position this at the bottom (Y-axis)
        text=str(count),  # Show the count value
        showarrow=False,
        font=dict(size=12, color="black"),
        align="center",
        # Adjust the vertical positioning so labels don't overlap
        yshift=5 if gender == 'Female' else -5,
        opacity=0.8  # Make the label slightly transparent
    )

# Update the layout to improve the appearance
fig_box.update_layout(
    yaxis_title="Amount Spent",
    xaxis_title="Income Level",
    plot_bgcolor="white",
    title="Spending Outliers by Income Level",
    showlegend=True,
    legend_title="Gender"
)

# Show the plot
fig_box.show()
```
