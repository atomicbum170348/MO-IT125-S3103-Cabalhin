# Who are our customers?

```python
# Set aesthetics
# Set default colormap to 'Greens'
plt.rcParams['image.cmap'] = 'Greens'
custom_green_palette = sns.color_palette("Greens", n_colors=10)
sns.set_theme(style="whitegrid", palette=custom_green_palette)
```

```python
# Ensure the SignupDate column is in datetime format
final_combined_dataset['SignupDate'] = pd.to_datetime(final_combined_dataset['SignupDate'], errors='coerce')

# Create a copy of the DataFrame to avoid the SettingWithCopyWarning
df_copy = final_combined_dataset.copy()

# Extract Year from SignupDate
df_copy['Year'] = df_copy['SignupDate'].dt.year

# Group by Year, counting unique CustomerIDs
signup_count = df_copy.groupby('Year').agg({'CustomerID': 'nunique'}).reset_index()

# Determine the color gradient based on the count of unique CustomerID
max_count = signup_count['CustomerID'].max()
min_count = signup_count['CustomerID'].min()

# Normalize the color scale: highest count gets the darkest, lowest count gets the lightest
signup_count['Color'] = signup_count['CustomerID'].apply(
    lambda x: (x - min_count) / (max_count - min_count)  # Normalized scale between 0 and 1
)

# Plotting
fig = px.bar(
    signup_count,
    x='CustomerID',
    y='Year',
    orientation='h',  # Horizontal bars
    title="Customer Signups per Year",
    labels={'CustomerID': 'Number of Signups', 'Year': 'Year'},
    color='Color',  # Color by normalized count
    color_continuous_scale='Greens',  # Use a green gradient where darkest = highest count
    text='CustomerID',  # Adding count annotations on the bars
)

# Remove color from tooltip
fig.update_traces(hovertemplate='Year: %{y}<br>Signups: %{x}<extra></extra>')  # Custom hover text

# Update layout
fig.update_layout(
    plot_bgcolor='white',
    xaxis_title="Number of Signups",
    yaxis_title="Year",
    showlegend=False,  # Hide the legend for a cleaner look
)

# Show the figure
fig.show()
```

```python
# Ensure the SignupDate column is in datetime format
final_combined_dataset = final_combined_dataset.copy()  # Make sure we're working on a copy

final_combined_dataset['SignupDate'] = pd.to_datetime(final_combined_dataset['SignupDate'], errors='coerce')

# Extract Year and Month
final_combined_dataset['Year'] = final_combined_dataset['SignupDate'].dt.year
final_combined_dataset['Month'] = final_combined_dataset['SignupDate'].dt.month

# Group by Year and Month, counting unique CustomerIDs
signup_count = final_combined_dataset.groupby(['Year', 'Month']).agg({'CustomerID': 'nunique'}).reset_index()

# Sort years in descending order
unique_years = sorted(signup_count['Year'].unique(), reverse=True)

# Generate reversed shades of green based on the number of years
green_shades = px.colors.sequential.Greens[len(unique_years):0:-1]  # Darkest shade for the most recent year

# Map years to specific green shades
year_to_color = {year: green_shades[i] for i, year in enumerate(unique_years)}

# Plotting
fig = px.line(
    signup_count,
    x='Month',
    y='CustomerID',
    color='Year',
    title="Customer Signups Over Time",
    labels={'CustomerID': 'Number of Signups', 'Month': 'Month'},
    markers=True,
    color_discrete_map=year_to_color  # Explicitly assign colors
)

fig.update_layout(
    plot_bgcolor='white',
    xaxis=dict(
        tickmode='array',
        tickvals=list(range(1, 13)),  # Ensure months 1-12 are shown
        ticktext=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
    ),
    yaxis_title="Number of Signups",
    xaxis_title="Month"
)
fig.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Example: Grouping data by Age with observed=False to suppress the FutureWarning
age_groups = pd.cut(final_combined_dataset['Age'], bins=[10, 19, 29, 39, 49, 59, 69, 79, 89, 100], labels=['10-19', '20-29', '30-39', '40-49', '50-59', '60-69', '70-79', '80-89', '90-100'])
age_count = final_combined_dataset.groupby(age_groups, observed=False)['CustomerID'].nunique().reset_index()  # added observed=False

# Plotting
plt.figure(figsize=(8, 6))
ax = sns.barplot(x='Age', y='CustomerID', data=age_count, palette="Greens", hue='Age', legend=False)  # used hue='Age' and legend=False

# Apply dynamic color shading based on the highest count
norm = plt.Normalize(age_count['CustomerID'].min(), age_count['CustomerID'].max())
sm = plt.cm.ScalarMappable(cmap="Greens", norm=norm)
for p in ax.patches:
    p.set_facecolor(sm.to_rgba(p.get_height()))

# Adding data point labels
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                fontsize=12, color='black', 
                xytext=(0, 5), textcoords='offset points')

plt.title('Age Demographics')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.show()
```

```python
# Grouping by IncomeLevel
income_count = final_combined_dataset.groupby('IncomeLevel')['CustomerID'].nunique().reset_index()

# Sorting by IncomeLevel in the specified order: Low, Medium, High
income_count['IncomeLevel'] = pd.Categorical(income_count['IncomeLevel'], categories=['Low', 'Medium', 'High'], ordered=True)
income_count = income_count.sort_values('IncomeLevel')

# Plotting
plt.figure(figsize=(8, 6))
ax = sns.barplot(x='IncomeLevel', y='CustomerID', data=income_count)

# Apply dynamic color shading based on the highest count, making darker as the count goes up
norm = plt.Normalize(income_count['CustomerID'].min(), income_count['CustomerID'].max())
sm = plt.cm.ScalarMappable(cmap="Greens", norm=norm)  # Use 'Greens' colormap
sm.set_array([])  # Set an empty array for ScalarMappable

for p in ax.patches:
    # Set the color of each bar according to its height (CustomerID count)
    p.set_facecolor(sm.to_rgba(p.get_height()))

# Adding data point labels
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                fontsize=12, color='black', 
                xytext=(0, 5), textcoords='offset points')  # Adjust xytext for better positioning

plt.title('Customer Spending Potential')
plt.xlabel('Income Level')
plt.ylabel('Number of Customers')

# Add colorbar to visualize color intensity
plt.colorbar(sm, ax=ax, label='Customer Count')  # Associate colorbar with the Axes and ScalarMappable

plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Assuming the final_combined_dataset already has the 'Age' and 'Amount' columns
# Create age bins (10-year intervals)
age_bins = pd.cut(final_combined_dataset['Age'], bins=range(0, 101, 10), right=False, labels=["0-9", "10-19", "20-29", "30-39", "40-49", "50-59", "60-69", "70-79", "80-89", "90-99"])

# Use .loc to avoid the SettingWithCopyWarning
final_combined_dataset.loc[:, 'AgeBin'] = age_bins

# Ensure the 'Amount' column is numeric
final_combined_dataset.loc[:, 'Amount'] = pd.to_numeric(final_combined_dataset['Amount'], errors='coerce')

# Group by AgeBin and calculate the total spending (sum of Amount) for each bin
age_bin_spending = final_combined_dataset.groupby('AgeBin', observed=True).agg({'Amount': 'sum'}).reset_index()

# Plotting the bar chart
fig = px.bar(
    age_bin_spending,
    x='AgeBin',
    y='Amount',
    title="Total Spending by Age Group",
    labels={'Amount': 'Total Spending', 'AgeBin': 'Age Group'},
    color='Amount',  # Color by total spending amount
    color_continuous_scale='Greens',  # Using a green gradient where darkest = highest spending
    text='Amount',  # Add total spending as annotations on the bars
)

# Update layout to improve readability
fig.update_traces(hovertemplate='Age Group: %{x}<br>Spending: %{y}<extra></extra>')  # Custom hover text

# Update layout
fig.update_layout(
    plot_bgcolor='white',
    xaxis_title="Age Group",
    yaxis_title="Total Spending",
    showlegend=False,  # Hide the legend for a cleaner look
)

# Show the figure
fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Ensure 'Amount' is numeric
final_combined_dataset['Amount'] = pd.to_numeric(final_combined_dataset['Amount'], errors='coerce')

# Group by ProductCategory and Gender and sum the Amount
gender_spending_by_category = final_combined_dataset.groupby(['ProductCategory', 'Gender']).agg({'Amount': 'sum'}).reset_index()

# Calculate total spending per ProductCategory
total_spending_by_category = (
    gender_spending_by_category.groupby('ProductCategory')['Amount'].sum().reset_index()
)
total_spending_by_category.rename(columns={'Amount': 'Total_Amount'}, inplace=True)

# Merge total spending data with gender spending data
gender_spending_by_category = gender_spending_by_category.merge(total_spending_by_category, on='ProductCategory')

# Calculate the percentage for each gender-product category combination
gender_spending_by_category['Percentage'] = (gender_spending_by_category['Amount'] / gender_spending_by_category['Total_Amount']) * 100

# Create the grouped bar chart
fig = px.bar(
    gender_spending_by_category,
    x='ProductCategory',
    y='Amount',
    color='Gender',
    barmode='group',  # Group bars by gender for each product category
    title="Spending by Gender for Each Product Category",
    labels={"ProductCategory": "Product Category", "Amount": "Total Spending"},
    color_discrete_sequence=['#006400', '#90EE90'],  # Dark green for Male, Light green for Female
)

# Add annotations inside each Gender's bar with amount and percentage
for i, row in gender_spending_by_category.iterrows():
    # Calculate x and y positions
    x_position = row['ProductCategory']
    if row['Gender'] == 'Female':
        offset = -0.5  # Adjust Female bar position
    else:
        offset = 0.5  # Adjust Male bar position
    
    # Create the text to display
    text = f'{row["Amount"]:.2f}<br>({row["Percentage"]:.1f}%)'  # Amount and Percentage

    fig.add_annotation(
        x=x_position,
        y=row['Amount'] / 1.3,  # Center the annotation inside the bar
        text=text,  # Display the amount and percentage
        showarrow=False,
        font=dict(color='white', size=10),  # Set font color to white for contrast
        align='center',
        xanchor='center',  # Center text horizontally
        xshift=offset * 100  # Slightly shift annotations to align perfectly
    )

# Add total annotations outside the bars
for i, row in total_spending_by_category.iterrows():
    fig.add_annotation(
        x=row['ProductCategory'],  # Place annotation above the bars
        y=row['Total_Amount'] + 5,  # Position outside the tallest bar
        text=f'Total: {row["Total_Amount"]:.2f}',  # Display the total amount with two decimals
        showarrow=False,
        font=dict(color='black', size=12),  # Set the font color for visibility
        align='center',
        xanchor='center',  # Center the annotation horizontally
    )

# Update layout for better readability
fig.update_layout(
    plot_bgcolor='white',
    xaxis_title='Product Category',
    yaxis_title='Total Spending',
    showlegend=True,  # Show the legend for gender
    xaxis_tickangle=0,  # Rotate x-axis labels for better visibility
    yaxis=dict(
        showgrid=True,  # Show gridlines for better readability
        tickmode='linear',  # Use linear ticks
        tick0=0,  # Start the ticks from 0
        dtick=50000,  # Set the tick step to 50,000
    ),
    font=dict(size=12)  # Adjust font size for clarity
)

# Show the figure
fig.show()


# Ensure 'TransactionID' is not null
final_combined_dataset['TransactionID'] = final_combined_dataset['TransactionID'].fillna('')

# Group by ProductCategory and Gender and count unique TransactionID
gender_transaction_count_by_category = final_combined_dataset.groupby(['ProductCategory', 'Gender']).agg({'TransactionID': pd.Series.nunique}).reset_index()

# Calculate total unique TransactionID count per ProductCategory
total_transaction_count_by_category = (
    gender_transaction_count_by_category.groupby('ProductCategory')['TransactionID'].sum().reset_index()
)
total_transaction_count_by_category.rename(columns={'TransactionID': 'Total_TransactionCount'}, inplace=True)

# Merge total transaction count data with gender transaction count data
gender_transaction_count_by_category = gender_transaction_count_by_category.merge(total_transaction_count_by_category, on='ProductCategory')

# Calculate the percentage for each gender-product category combination
gender_transaction_count_by_category['Percentage'] = (gender_transaction_count_by_category['TransactionID'] / gender_transaction_count_by_category['Total_TransactionCount']) * 100

# Create the grouped bar chart
fig = px.bar(
    gender_transaction_count_by_category,
    x='ProductCategory',
    y='TransactionID',
    color='Gender',
    barmode='group',  # Group bars by gender for each product category
    title="Transactions by Gender for Each Product Category",
    labels={"ProductCategory": "Product Category", "TransactionID": "Count of Unique TransactionID"},
    color_discrete_sequence=['#006400', '#90EE90'],  # Dark green for Male, Light green for Female
)

# Add annotations inside each Gender's bar with count and percentage
for i, row in gender_transaction_count_by_category.iterrows():
    # Calculate x and y positions
    x_position = row['ProductCategory']
    if row['Gender'] == 'Female':
        offset = -0.5  # Adjust Female bar position
    else:
        offset = 0.5  # Adjust Male bar position
    
    # Create the text to display
    text = f'{row["TransactionID"]}<br>({row["Percentage"]:.1f}%)'  # Count and Percentage

    fig.add_annotation(
        x=x_position,
        y=row['TransactionID'] / 1.3,  # Center the annotation inside the bar
        text=text,  # Display the count and percentage
        showarrow=False,
        font=dict(color='white', size=10),  # Set font color to white for contrast
        align='center',
        xanchor='center',  # Center text horizontally
        xshift=offset * 100  # Slightly shift annotations to align perfectly
    )

# Add total annotations outside the bars
for i, row in total_transaction_count_by_category.iterrows():
    fig.add_annotation(
        x=row['ProductCategory'],  # Place annotation above the bars
        y=row['Total_TransactionCount'] + 5,  # Position outside the tallest bar
        text=f'Total: {row["Total_TransactionCount"]}',  # Display the total count of unique TransactionID
        showarrow=False,
        font=dict(color='black', size=12),  # Set the font color for visibility
        align='center',
        xanchor='center',  # Center the annotation horizontally
    )

# Update layout for better readability
fig.update_layout(
    plot_bgcolor='white',
    xaxis_title='Product Category',
    yaxis_title='Count of Transactions',
    showlegend=True,  # Show the legend for gender
    xaxis_tickangle=0,  # Rotate x-axis labels for better visibility
    yaxis=dict(
        showgrid=True,  # Show gridlines for better readability
        tickmode='linear',  # Use linear ticks
        tick0=0,  # Start the ticks from 0
        dtick=50,  # Set the tick step to 50
    ),
    font=dict(size=12)  # Adjust font size for clarity
)

# Show the figure
fig.show()
```
